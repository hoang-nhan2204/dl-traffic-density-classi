# Deep Custom CNN for Traffic Density Classification

This notebook evaluates a Deep CNN trained from scratch to classify traffic images into five density levels. The experiments compare the effects of training-time augmentation and class weights while keeping the model architecture and data pipeline consistent.


```python
import random
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras import layers, models
```

```python
project_dir = Path.cwd()
if not (project_dir / "src").exists():
    project_dir = project_dir.parent
if str(project_dir) not in sys.path:
    sys.path.append(str(project_dir))

from src.data_loader import CLASS_NAMES, create_datasets
from src.evaluation import evaluate_model
```

## 1. Experiment Setup

```python
# Set shared values and random seeds for reproducible experiments
SEED = 42
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
STEPS_PER_EXECUTION = 1
EPOCHS = 20

random.seed(SEED)
np.random.seed(SEED)
tf.keras.utils.set_random_seed(SEED)

RUN_DIR = project_dir / "outputs" / "complex_cnn_notebook"
RUN_DIR.mkdir(parents=True, exist_ok=True)

print("Image size:", IMAGE_SIZE)
print("Batch size:", BATCH_SIZE)
print("Steps per execution:", STEPS_PER_EXECUTION)
print("Maximum epochs:", EPOCHS)
print("Outputs:", RUN_DIR)
print("Mixed precision:", tf.keras.mixed_precision.global_policy().name)
```

The experiments use a shared random seed, image size, and batch size. Early stopping monitors validation Macro F1 and stops training when this metric no longer improves.

## 2. Data Loading

```python
# Load the cleaned train, validation, and test splits from manifest files
train_dataset, val_dataset, test_dataset = create_datasets(
    augmentation=False,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
)

print("Classes:", CLASS_NAMES)
print("Training batches:", train_dataset.cardinality().numpy())
print("Validation batches:", val_dataset.cardinality().numpy())
print("Test batches:", test_dataset.cardinality().numpy())
```

The cleaned train, validation, and test splits are loaded from manifest files through the `tf.data.Dataset` pipeline. Augmentation is applied only to the training split in experiments that enable it; validation and test data remain unchanged for fair evaluation.

## 3. Class Distribution and Class Weights

```python
# Compute class weights using only the training labels
train_manifest = pd.read_csv(project_dir / "data" / "manifests" / "train.csv")
class_to_index = {name: index for index, name in enumerate(CLASS_NAMES)}
train_counts = train_manifest["class_name"].value_counts().reindex(CLASS_NAMES, fill_value=0)
train_class_ids = train_manifest["class_name"].map(class_to_index).to_numpy()
class_ids = np.arange(len(CLASS_NAMES))

class_weight_values = compute_class_weight(
    class_weight="balanced", classes=class_ids, y=train_class_ids,
)
class_weights = {
    int(class_id): float(weight)
    for class_id, weight in zip(class_ids, class_weight_values)
}
class_weight_table = pd.DataFrame({
    "Class": CLASS_NAMES,
    "Number of images": train_counts.to_numpy(),
    "Class weight": [class_weights[index] for index in class_ids],
})
class_weight_table
```

The class distribution is imbalanced, so Experiment 3 uses class weights to increase the loss contribution of underrepresented classes. The weights are calculated from training labels only and are used only during training.

## 4. Deep Custom CNN Architecture

```python
# Define the deeper custom CNN used in all three experiments
def build_deep_cnn():
    model = models.Sequential([
        layers.Input(shape=(*IMAGE_SIZE, 3)),
        layers.Conv2D(32, 3, padding="same", use_bias=False),
        layers.BatchNormalization(), layers.Activation("relu"),
        layers.Conv2D(32, 3, padding="same", use_bias=False),
        layers.BatchNormalization(), layers.Activation("relu"),
        layers.MaxPooling2D(2),
        layers.Conv2D(64, 3, padding="same", use_bias=False),
        layers.BatchNormalization(), layers.Activation("relu"),
        layers.Conv2D(64, 3, padding="same", use_bias=False),
        layers.BatchNormalization(), layers.Activation("relu"),
        layers.MaxPooling2D(2),
        layers.Conv2D(128, 3, padding="same", use_bias=False),
        layers.BatchNormalization(), layers.Activation("relu"),
        layers.Conv2D(128, 3, padding="same", use_bias=False),
        layers.BatchNormalization(), layers.Activation("relu"),
        layers.MaxPooling2D(2),
        layers.Conv2D(256, 3, padding="same", use_bias=False),
        layers.BatchNormalization(), layers.Activation("relu"),
        layers.MaxPooling2D(2),
        layers.GlobalAveragePooling2D(),
        layers.Dense(128, activation="relu"),
        layers.Dropout(0.4),
        layers.Dense(len(CLASS_NAMES), activation="softmax", dtype="float32"),
    ])
    return model

model_preview = build_deep_cnn()
model_preview.summary()
M1_PARAMETERS = 2_792_869
print(f"M1 parameters: {M1_PARAMETERS:,}")
print(f"M2 parameters: {model_preview.count_params():,}")
```

Compared with the Simple CNN in M1, the Deep CNN uses more convolutional blocks to learn features at different levels. Batch normalization supports stable learning, global average pooling reduces the number of classifier parameters compared with flattening, and dropout helps reduce overfitting.

## 5. Training and Visualization Utilities

```python
# Initialize experiment tracking once before Experiments 1, 2, and 3.
experiment_models = {}
experiment_histories = {}
experiment_records = []
```

```python
# Track validation Macro F1 after each epoch for model selection
class ValidationMacroF1(tf.keras.callbacks.Callback):
    def __init__(self, dataset):
        super().__init__()
        self.dataset = dataset

    def on_epoch_end(self, epoch, logs=None):
        probabilities = self.model.predict(self.dataset, verbose=0)
        y_pred = np.argmax(probabilities, axis=1)
        y_true = get_true_labels(self.dataset)
        logs["val_macro_f1"] = f1_score(
            y_true, y_pred, average="macro", zero_division=0,
        )


def make_callbacks(val_dataset, checkpoint_path):
    # Selection, checkpointing and early stopping all use the same validation metric.
    return [
        ValidationMacroF1(val_dataset),
        tf.keras.callbacks.ModelCheckpoint(
            checkpoint_path,
            monitor="val_macro_f1",
            mode="max",
            save_best_only=True,
        ),
        tf.keras.callbacks.EarlyStopping(
            monitor="val_macro_f1",
            mode="max",
            patience=6,
            restore_best_weights=True,
            start_from_epoch=5,
        ),
        tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_macro_f1",
            mode="max",
            factor=0.5,
            patience=3,
            min_lr=1e-6,
            verbose=1,
        ),
    ]


def plot_history(history, title):
    epochs = range(1, len(history.history["loss"]) + 1)
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(epochs, history.history["loss"], label="Training loss")
    plt.plot(epochs, history.history["val_loss"], label="Validation loss")
    plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title(f"{title} - Loss"); plt.legend()
    plt.subplot(1, 2, 2)
    plt.plot(epochs, history.history["accuracy"], label="Training accuracy")
    plt.plot(epochs, history.history["val_accuracy"], label="Validation accuracy")
    plt.xlabel("Epoch"); plt.ylabel("Accuracy"); plt.title(f"{title} - Accuracy"); plt.legend()
    plt.tight_layout()
    plt.show()


def get_true_labels(dataset):
    return np.concatenate([np.argmax(labels.numpy(), axis=1) for _, labels in dataset])


def get_validation_macro_f1(model, dataset):
    probabilities = model.predict(dataset, verbose=0)
    predicted_labels = np.argmax(probabilities, axis=1)
    return f1_score(
        get_true_labels(dataset), predicted_labels, average="macro", zero_division=0,
    )


def record_experiment(name, history, model, val_dataset, augmentation, weighted):
    experiment_models[name] = model
    experiment_histories[name] = history
    experiment_records.append({
        "Experiment": name,
        "Augmentation": augmentation,
        "Class Weights": weighted,
        "Best Val Accuracy": max(history.history["val_accuracy"]),
        "Best Val Macro F1": get_validation_macro_f1(model, val_dataset),
        "Lowest Val Loss": min(history.history["val_loss"]),
        "Epochs Trained": len(history.history["loss"]),
    })
    result = experiment_records[-1]
    print(f"Best validation accuracy: {result['Best Val Accuracy']:.4f}")
    print(f"Validation macro F1: {result['Best Val Macro F1']:.4f}")
    print(f"Lowest validation loss: {result['Lowest Val Loss']:.4f}")
```

## 6. Experiment 1: Deep CNN Baseline

This experiment trains the deeper custom CNN without augmentation or class weights. It measures the effect of the improved architecture relative to the Simple CNN from M1.


```python
# Train a fresh baseline model without augmentation or class weights
train_dataset_1, val_dataset_1, _ = create_datasets(
    augmentation=False, image_size=IMAGE_SIZE, batch_size=BATCH_SIZE,
)
model_exp1 = build_deep_cnn()
model_exp1.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy", metrics=["accuracy"],
    steps_per_execution=STEPS_PER_EXECUTION,
)
history_exp1 = model_exp1.fit(
    train_dataset_1, validation_data=val_dataset_1, epochs=EPOCHS,
    callbacks=make_callbacks(val_dataset_1, RUN_DIR / "exp1_best.keras"),
)
plot_history(history_exp1, "Experiment 1: Deep CNN baseline")
record_experiment(
    "Deep CNN baseline", history_exp1, model_exp1, val_dataset_1,
    augmentation=False, weighted=False,
)
```

## 7. Experiment 2: Deep CNN + Augmentation

The architecture and training settings remain the same as in Experiment 1, while training-time augmentation is enabled. A new model is initialized and trained from scratch; training does not continue from Experiment 1.


```python
# Train a fresh model with training-time augmentation enabled
train_dataset_2, val_dataset_2, _ = create_datasets(
    augmentation=True, image_size=IMAGE_SIZE, batch_size=BATCH_SIZE,
)
model_exp2 = build_deep_cnn()
model_exp2.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy", metrics=["accuracy"],
    steps_per_execution=STEPS_PER_EXECUTION,
)
history_exp2 = model_exp2.fit(
    train_dataset_2, validation_data=val_dataset_2, epochs=EPOCHS,
    callbacks=make_callbacks(val_dataset_2, RUN_DIR / "exp2_best.keras"),
)
plot_history(history_exp2, "Experiment 2: Deep CNN + Augmentation")
record_experiment(
    "Deep CNN + Augmentation", history_exp2, model_exp2, val_dataset_2,
    augmentation=True, weighted=False,
)
```

## 8. Experiment 3: Deep CNN + Augmentation + Class Weights

This experiment adds class weights to the augmented setup to investigate whether emphasizing minority classes improves class-balanced performance. The weights are calculated using only the training split. A new model is initialized and trained from scratch.


```python
# Train a fresh augmented model using class weights from the training split
train_dataset_3, val_dataset_3, _ = create_datasets(
    augmentation=True, image_size=IMAGE_SIZE, batch_size=BATCH_SIZE,
)
model_exp3 = build_deep_cnn()
model_exp3.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy", metrics=["accuracy"],
    steps_per_execution=STEPS_PER_EXECUTION,
)
history_exp3 = model_exp3.fit(
    train_dataset_3, validation_data=val_dataset_3, epochs=EPOCHS,
    callbacks=make_callbacks(val_dataset_3, RUN_DIR / "exp3_best.keras"), class_weight=class_weights,
)
plot_history(history_exp3, "Experiment 3: Deep CNN + Augmentation + Class Weights")
record_experiment(
    "Deep CNN + Augmentation + Class Weights", history_exp3, model_exp3, val_dataset_3,
    augmentation=True, weighted=True,
)
```

## 9. Experiment Comparison and Final Configuration

The experiments are compared primarily using validation metrics, especially Macro F1 because the dataset is imbalanced. The selected configuration is the one with the best validation Macro F1, with accuracy and loss used as tie-breakers.


```python
# Rank experiments by validation Macro F1, then accuracy and loss
experiment_results = pd.DataFrame(experiment_records)
experiment_results = experiment_results.sort_values(
    by=["Best Val Macro F1", "Best Val Accuracy", "Lowest Val Loss"],
    ascending=[False, False, True],
    ignore_index=True,
)
experiment_results
```

```python
selected_experiment = max(
    experiment_records,
    key=lambda row: (
        row["Best Val Macro F1"], row["Best Val Accuracy"], -row["Lowest Val Loss"]
    ),
)
final_model = experiment_models[selected_experiment["Experiment"]]
print("Selected final configuration:", selected_experiment["Experiment"])
print("Selection uses validation Macro F1, then accuracy and loss as tie-breakers.")
```

## 10. M2 Test-Set Evaluation

The selected M2 model is evaluated on the held-out test split. Accuracy summarizes overall classification performance. Macro Precision gives equal importance to each class when averaging precision, Macro Recall does the same for recall, and Macro F1 summarizes class-balanced precision and recall.


```python
# Evaluate the selected M2 configuration on the held-out test split
m2_result = evaluate_model(
    model=final_model,
    test_dataset=test_dataset,
    model_name="Deep Custom CNN",
)
m2_result
```

The confusion matrix shows how often examples from each true class are assigned to each predicted class. It helps identify which traffic-density classes are most frequently confused; the classification report summarizes per-class test performance.

```python
# Inspect test-set errors with a confusion matrix and per-class report
if final_model is not None:
    pred_probs = final_model.predict(test_dataset, verbose=0)
    y_pred = np.argmax(pred_probs, axis=1)
    y_true = get_true_labels(test_dataset)

    cm = confusion_matrix(y_true, y_pred, labels=np.arange(len(CLASS_NAMES)))
    plt.figure(figsize=(7, 6))
    plt.imshow(cm, interpolation="nearest", cmap="Blues")
    plt.title("Confusion Matrix - Deep Custom CNN")
    plt.colorbar()
    tick_positions = np.arange(len(CLASS_NAMES))
    plt.xticks(tick_positions, CLASS_NAMES, rotation=35, ha="right")
    plt.yticks(tick_positions, CLASS_NAMES)
    plt.xlabel("Predicted class")
    plt.ylabel("True class")
    threshold = cm.max() / 2 if cm.size else 0
    for row in range(cm.shape[0]):
        for column in range(cm.shape[1]):
            plt.text(column, row, format(cm[row, column], "d"),
                     ha="center", va="center",
                     color="white" if cm[row, column] > threshold else "black")
    plt.tight_layout()
    plt.show()

    print(classification_report(
        y_true,
        y_pred,
        target_names=CLASS_NAMES,
        digits=4,
        zero_division=0,
    ))
else:
    print("Confusion matrix and classification report skipped because the selected model is unavailable.")
```

## 13. Final Observations

On validation data, Experiment 1 performed best, with Macro F1 of 0.7984, compared with 0.7196 for Experiment 2 and 0.6970 for Experiment 3. In these runs, augmentation reduced validation performance, and adding class weights to the augmented setup did not improve validation Macro F1. The stored results do not include per-class validation metrics, so they do not establish whether class weighting improved minority-class performance.

The selected M2 model is the Experiment 1 baseline. On the test set, it achieved 0.8038 accuracy and 0.7957 Macro F1 (Macro Precision: 0.7829; Macro Recall: 0.8184). M2 outperformed the M1 Simple CNN on the reported test metrics: M1 achieved 0.7608 accuracy and 0.7582 F1-score. Augmentation and class weights do not always improve results: the original dataset may be sufficient for the baseline, some augmented variations may not help this task, and class weighting can improve attention to minority classes while reducing overall decision stability.
